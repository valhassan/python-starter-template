# How I would setup my enviroment for ML in 2025

The primary tools


- Python: This remains the programming language of choice of Machine Learning Scientists.

- UV: Super fast python enviroment manager built on Rust.

- Ruff: Super fast python linter and code formatter also written in Rust.

- Git and Github: Version control and cloud code repository (Public and Private).

- IDE: Development enviroment that greatly reduces frustration when writing your code. Feel free to use VIM or any editor, I admire your courage.



- Pytorch: Deep learning framework that provides routines for manipulating tensors (the holy grail of machine learning) + math operations.

- Lightning: Adds lightning and thunder to pytorch; provides high level abstraction to pytorch making your life somewhat easier.
